## Why transpose happens a lot in NN?
- **Forward:** each row of \(W\) tells us how the inputs combine to produce one output.
- **Backward:** we need to collect how all outputs affect each input, so we look at the same connections in the opposite direction — that’s why \(W\) becomes \(W^T\).

### Backprop from First Principles

At the lowest level, backpropagation is just the chain rule applied repeatedly over a computation graph. Even if a neural network uses vectors, matrices, and tensors, you can still think of everything as many scalar variables connected by simple operations. Each path contributes `upstream gradient × local derivative`, and if a variable affects the result through multiple paths, those contributions are added together.

Jacobians, matrix notation, and VJPs **do not introduce a new underlying mechanism**. They are higher-level abstractions for organizing this same process. Textbooks emphasize them because once models contain thousands or millions of variables, writing every partial derivative separately becomes impractical; these concepts provide a **compact, general mathematical language** for the same underlying chain-rule computation.

# when autograd stops being enough
**Context:** Matrix calculus is mostly an abstraction over ordinary multivariable calculus + chain rule. You usually don’t need to go deep unless the **mathematical structure itself becomes the research problem**.

- **Using existing models / layers** → autograd is enough
- **Standard training / backprop** → autograd is enough
- **Designing new losses / layers** → useful
- **Optimization research** → important
- **Hessians / curvature / loss landscape** → important
- **Jacobians / stability / sensitivity** → important
- **Second-order methods / preconditioning** → important
- **Probabilistic ML / covariance matrices** → useful
- **Normalizing flows / log-det Jacobians** → core
- **Meta-learning / bilevel optimization / implicit differentiation** → core
- **VJP / JVP / autodiff research** → core

**Mental model:**  
**Using the math → autograd**  
**Reasoning about / redesigning the math → matrix calculus**
