## ID, One-hot, and Embedding

**ID answers:**

> Which one is it?

**Embedding answers:**

> How should the model represent it?

---

An integer ID and a one-hot vector are two different ways of encoding **identity**.

For example:

```text
Integer ID:

A = 0
B = 1
C = 2
```

or:

```text
One-hot:

A = [1, 0, 0]
B = [0, 1, 0]
C = [0, 0, 1]
```

Neither representation tells us anything about what A, B, or C **means**.

They only tell us which one it is.

One-hot is useful because, unlike directly using integer IDs as numerical values, it does not introduce arbitrary numerical relationships such as:

```text
A < B < C
```

However, we do not actually need to construct the one-hot vector.

Mathematically:

```text
one-hot × embedding matrix
```

simply selects one row from the embedding matrix.

So instead of:

```text
token
→ token ID
→ one-hot
→ matrix multiplication
→ embedding
```

modern implementations can directly do:

```text
token
→ token ID
→ embedding lookup
```

Therefore:

> **One-hot is mathematically useful, but computationally unnecessary.**

A useful distinction is:

```text
ID / one-hot:
Which one?

Embedding:
How should it be represented?
```
