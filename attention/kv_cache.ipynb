## KV Cache

> There are 3 tokens, now, next token comes in, will it affect Q, K, V of the first 3 tokens?
> 
> No!



> Remember Causal mask:
>
> For example, will the 3rd token grab ANY info from the 4th token?
>
> No! it can't look into the FUTURE!
>
> So adding token 4 won't change token 3's state


### What to cache when the 4th token comes in?

> To calculate the 4th token's attention output:
>
> - We need **Q4** to query K1–4 and retrieve information from V1–4.
> - K1–3 and V1–3 are unchanged → **Cache them!**
> - Q1–3 are no longer needed → **No need to cache them!**
> - Q4, K4, V4 are calculated from the new input.
>
> So we cache the historical K and V **at every Transformer layer**. That's why it's called **KV Cache**.
