## What changed using multi-head?

**More softmax! More independent ways to gather contextual information!**

Each head can learn to focus on a different kind of context, using its own attention distribution.

### Why can `d_head` be smaller than `d_model`?

Because understanding a particular context often doesn't require all the information in the original representation.

Think of visiting a doctor.

You only have 20 minutes. You could spend those 20 minutes telling the doctor your entire life story, or focus on the information relevant to the medical problem.

A cardiologist and a neurologist might need very different information from the same patient.

Similarly, each attention head receives the full input representation but learns a smaller, potentially lossy projection that can preserve information relevant to its own attention pattern.

Each head has limited capacity, but that's okay: **not every head needs to preserve everything.**

### Why multi-head?

Instead of using our entire representation budget for one attention pattern, we split it across multiple heads.

Each head can learn a different way to select and mix contextual information.

Finally, we concatenate the head outputs and use \(W_O\) to combine them into a `d_model`-dimensional representation.

**The intuition: Same representation budget, multiple specialized views. Potentially better use of limited capacity.**

These specializations are learned, not predefined or guaranteed.
