```python
import numpy as np

# -------------------------
# Scalar encoding
# -------------------------
A_num = np.array([1.0])
B_num = np.array([2.0])
C_num = np.array([3.0])

print("Scalar encoding distances:")
print("A-B:", np.linalg.norm(A_num - B_num))
print("B-C:", np.linalg.norm(B_num - C_num))
print("A-C:", np.linalg.norm(A_num - C_num))


# -------------------------
# One-hot encoding
# -------------------------
A_hot = np.array([1.0, 0.0, 0.0])
B_hot = np.array([0.0, 1.0, 0.0])
C_hot = np.array([0.0, 0.0, 1.0])

print("\nOne-hot encoding distances:")
print("A-B:", np.linalg.norm(A_hot - B_hot))
print("B-C:", np.linalg.norm(B_hot - C_hot))
print("A-C:", np.linalg.norm(A_hot - C_hot))
```
